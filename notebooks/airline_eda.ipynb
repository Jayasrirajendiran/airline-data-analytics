# Airline Data Analytics - Comprehensive Exploratory Data Analysis (EDA)

**Dataset**: Flight-level profitability & operational records (7,974 flights)
**Objective**: Evaluate revenue, profitability, fleet performance, route economics, cost drivers, and seasonal demand.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

BASE_DIR = Path('.').resolve().parent if Path('.').resolve().name == 'notebooks' else Path('.').resolve()
DATA_FILE = BASE_DIR / 'data' / 'processed' / 'airline_cleaned.csv'
LOOKUP_FILE = BASE_DIR / 'data' / 'lookup' / 'airport_lookup.csv'

df = pd.read_csv(DATA_FILE)
airports = pd.read_csv(LOOKUP_FILE)
df['Flight_Date'] = pd.to_datetime(df['Flight_Date'])
print('Dataset loaded successfully with shape:', df.shape)
```

## 1. Executive Business KPIs & Validation Checks

```python
kpis = {
    'Total Flights': len(df),
    'Total Passengers': df['Passengers'].sum(),
    'Total Revenue ($)': df['Total_Revenue'].sum(),
    'Total Cost ($)': df['Total_Cost'].sum(),
    'Total Profit ($)': df['Profit'].sum(),
    'Overall Profit Margin (%)': (df['Profit'].sum() / df['Total_Revenue'].sum()) * 100
}
pd.DataFrame(list(kpis.items()), columns=['KPI Metric', 'Calculated Value'])
```
